```python
# --- Dhruv Patel Academy practice checker (runs in YOUR notebook, on YOUR variables) ---
import json as _json, sys as _sys, urllib.request as _ur

class _DSQ:
    def __init__(self):
        self.qs = {}; self.order = []; self.tries = {}; self.ok = {}; self.url = None; self.key = None; self.module = ''

    def load(self, pack):
        self.module = pack['title']
        for q in pack['questions']:
            self.qs[q['id']] = q; self.order.append(q['id'])
        print(f"Dhruv Patel Academy: {len(self.order)} questions loaded for '{self.module}'. Write your answer, then run the check cell under it.")
        if self._auto(): print('Linked to your Dhruv Patel Academy account: solved questions count in the app.')

    def connect(self, url, key, quiet=False):
        self.url = url.rstrip('/'); self.key = key
        if not quiet: print('Connected: solved questions will be reported to', self.url)

    def _auto(self):
        # Dhruv Patel Academy writes dsq_config.json next to the notebooks when you open them from the app
        import os
        d = os.getcwd()
        for _ in range(4):
            p = os.path.join(d, 'dsq_config.json')
            if os.path.exists(p):
                try:
                    c = _json.load(open(p)); self.connect(c['url'], c['key'], quiet=True); return True
                except Exception:
                    pass
            d = os.path.dirname(d)
        return False

    def _report(self, qid, ok):
        if not self.url: return
        try:
            req = _ur.Request(self.url + '/api/nb/report', data=_json.dumps({'key': self.key, 'qid': qid, 'ok': bool(ok)}).encode(), headers={'Content-Type': 'application/json'})
            _ur.urlopen(req, timeout=4).read()
        except Exception as e:
            print('(could not reach the app:', str(e)[:80] + ')')

    def check(self, qid):
        q = self.qs[qid]
        try:
            ip = get_ipython(); ns = dict(ip.user_ns)
        except NameError:
            ns = dict(_sys.modules['__main__'].__dict__)
        results = []
        def check(cond, msg, hint=''):
            ok = bool(cond); results.append((ok, msg, hint)); return ok
        ns['check'] = check; ns['OUT'] = ''
        for mod, alias in (('numpy', 'np'), ('pandas', 'pd')):
            if alias not in ns:
                try: ns[alias] = __import__(mod)
                except Exception: pass
        if 'plt' not in ns:
            try:
                import matplotlib.pyplot as plt; ns['plt'] = plt
            except Exception: pass
        try:
            exec(compile(q['tests'], '<tests>', 'exec'), ns)
        except BaseException as e:
            results.append((False, 'A check could not finish: ' + type(e).__name__ + ': ' + str(e)[:200], 'Did you create every variable name the task asks for? Did you run your answer cell first?'))
        self.tries[qid] = self.tries.get(qid, 0) + 1
        good = bool(results) and all(r[0] for r in results)
        for ok, msg, hint in results:
            print(('  [PASS] ' if ok else '  [FAIL] ') + msg + ('' if ok or not hint else '\n         hint: ' + hint))
        if good:
            self.ok[qid] = True; print(f"\n  Question solved: {q['title']}  (attempt {self.tries[qid]})")
        else:
            print(f"\n  Not yet - fix your answer cell, run it again, then re-run this check. (attempt {self.tries[qid]})")
        self._report(qid, good)
        return good

    def hint(self, qid):
        hs = self.qs[qid].get('hints') or []
        n = min(self.tries.get(qid, 0), len(hs)) if hs else 0
        print(('hint: ' + hs[max(0, n - 1)]) if hs else 'No hint for this one - re-read the question.')

    def solution(self, qid):
        if self.tries.get(qid, 0) < 3 and not self.ok.get(qid):
            print('Try the check at least 3 times first - the struggle is where the learning happens.'); return
        print(self.qs[qid]['solution'])

    def score(self):
        n = len(self.order); s = sum(1 for q in self.order if self.ok.get(q))
        print(f"Solved {s} of {n}.")
        for q in self.order:
            print(('  [x] ' if self.ok.get(q) else '  [ ] ') + q + ' - ' + self.qs[q]['title'] + ('' if self.ok.get(q) else f"  (tries: {self.tries.get(q, 0)})"))

dsq = _DSQ()


dsq.load({
 "module": "matplotlib",
 "title": "Matplotlib & Seaborn",
 "questions": [
  {
   "id": "mp-1-1",
   "title": "Your first labelled chart",
   "tests": "ax = plt.gcf().axes[0]\ncheck(len(ax.lines) == 1 and list(ax.lines[0].get_ydata()) == [120, 135, 160, 150, 190, 210], 'the line shows the sales data', 'ax.plot(months, sales)')\ncheck(len(ax.get_title()) > 3 and ax.get_xlabel() and ax.get_ylabel(), 'title and both axis labels are set', 'set_title / set_xlabel / set_ylabel')\nleg = ax.get_legend()\ncheck(leg is not None and [t.get_text() for t in leg.get_texts()] == ['2026'], 'the legend shows \"2026\"', 'label=\"2026\" and ax.legend()')",
   "hints": [
    "label= on plot gives the legend its text; ax.legend() draws the legend."
   ],
   "solution": "import matplotlib.pyplot as plt\n\nmonths = [1, 2, 3, 4, 5, 6]\nsales = [120, 135, 160, 150, 190, 210]\n\nfig, ax = plt.subplots(figsize=(7, 3.4))\nax.plot(months, sales, marker=\"o\", label=\"2026\")\nax.set_title(\"Monthly sales\")\nax.set_xlabel(\"month\")\nax.set_ylabel(\"sales\")\nax.legend()"
  },
  {
   "id": "mp-1-2",
   "title": "Two series, one axes",
   "tests": "ax = plt.gcf().axes[0]\ncheck(len(ax.lines) == 2, 'two lines are drawn', 'two ax.plot(...) calls')\ncheck(ax.lines[1].get_linestyle() == '--', 'the second line is dashed', 'linestyle=\"--\"')\ncheck(sorted(t.get_text() for t in ax.get_legend().get_texts()) == ['actual', 'plan'], 'legend shows plan and actual', 'label=... on each plot call')\ncheck(ax.get_title() and ax.get_xlabel() and ax.get_ylabel(), 'title and axis labels are set', 'set_title, set_xlabel, set_ylabel')\ncheck(any(l.get_visible() for l in ax.xaxis.get_gridlines()) and ax.xaxis._major_tick_kw.get('gridOn', False), 'a grid is on', 'ax.grid(alpha=0.3)')",
   "hints": [
    "Two ax.plot calls on the same ax.",
    "linestyle=\"--\" gives a dashed line."
   ],
   "solution": "import numpy as np\nimport matplotlib.pyplot as plt\n\nx = np.arange(1, 9)\na = x * 10\nb = x * 10 + np.array([2, -3, 4, -6, 5, -8, 9, -4])\n\nfig, ax = plt.subplots(figsize=(7, 3.4))\nax.plot(x, a, label=\"plan\")\nax.plot(x, b, linestyle=\"--\", label=\"actual\")\nax.set_title(\"Plan vs actual\")\nax.set_xlabel(\"week\")\nax.set_ylabel(\"units\")\nax.legend()\nax.grid(alpha=0.3)"
  },
  {
   "id": "mp-2-1",
   "title": "Scatter by group",
   "tests": "ax = plt.gcf().axes[0]\ncheck(len(ax.collections) == 2, 'two scatter groups are drawn', 'one ax.scatter per group')\ncheck(all(len(c.get_offsets()) == 40 for c in ax.collections), 'each group has 40 points', 'scatter(tx, ty) for the second group')\ncheck(sorted(t.get_text() for t in ax.get_legend().get_texts()) == ['control', 'treated'], 'legend shows control and treated', 'label=\"treated\" and ax.legend()')\ncheck(ax.get_xlabel() and ax.get_ylabel(), 'both axes are labelled', 'set_xlabel / set_ylabel')",
   "hints": [
    "The second call plots the treated points: tx, ty."
   ],
   "solution": "import numpy as np\nimport matplotlib.pyplot as plt\n\nrng = np.random.default_rng(1)\ncx, cy = rng.normal(2, 0.6, 40), rng.normal(3, 0.6, 40)\ntx, ty = rng.normal(4, 0.6, 40), rng.normal(5, 0.6, 40)\n\nfig, ax = plt.subplots(figsize=(6, 4))\nax.scatter(cx, cy, color=\"#2F6FE0\", alpha=0.7, label=\"control\")\nax.scatter(tx, ty, color=\"#D2600F\", alpha=0.7, label=\"treated\")\nax.set_xlabel(\"dose\")\nax.set_ylabel(\"response\")\nax.legend()"
  },
  {
   "id": "mp-2-2",
   "title": "Bars and a histogram",
   "tests": "fig = plt.gcf(); a1, a2 = fig.axes[0], fig.axes[1]\ncheck(len(a1.patches) == 4 and sorted(p.get_height() for p in a1.patches) == [190, 280, 320, 410], 'left chart has 4 bars with the sales values', 'ax1.bar(regions, sales)')\ncheck(len(a2.patches) == 12, 'right chart is a histogram with 12 bins', 'ax2.hist(scores, bins=12)')\ncheck(a1.get_title() == 'Sales by region' and a2.get_title() == 'Score distribution', 'both charts have the requested titles', 'set_title')\ncheck(a1.get_ylabel() and a2.get_ylabel(), 'both charts have a y label', 'set_ylabel')",
   "hints": [
    "bar(categories, heights)",
    "hist(data, bins=n) draws n bars."
   ],
   "solution": "import numpy as np\nimport matplotlib.pyplot as plt\n\nregions = [\"North\", \"South\", \"East\", \"West\"]\nsales = [320, 280, 410, 190]\nscores = np.random.default_rng(5).normal(60, 12, 300)\n\nfig, (ax1, ax2) = plt.subplots(1, 2, figsize=(9, 3.6))\nax1.bar(regions, sales)\nax1.set_title(\"Sales by region\")\nax1.set_ylabel(\"sales\")\nax2.hist(scores, bins=12)\nax2.set_title(\"Score distribution\")\nax2.set_ylabel(\"students\")\nfig.tight_layout()"
  },
  {
   "id": "mp-3-1",
   "title": "Annotate the peak",
   "tests": "ax = plt.gcf().axes[0]\ncheck(len(ax.lines) >= 2, 'the data line and an average line are drawn', 'ax.axhline(y.mean(), ...)')\nhl = [l for l in ax.lines if len(set(l.get_ydata())) == 1]\ncheck(hl and abs(hl[0].get_ydata()[0] - 221 / 12) < 1e-9, 'the reference line sits at the average (about 18.42)', 'ax.axhline(y.mean())')\ncheck(any(t.get_text() == 'peak' for t in ax.texts), 'there is an annotation \"peak\"', 'ax.annotate(\"peak\", ...)')\ncheck(ax.get_title() and ax.get_xlabel() and ax.get_ylabel(), 'title and axis labels are set', 'set_title / labels')",
   "hints": [
    "argmax gives the POSITION of the highest value."
   ],
   "solution": "import numpy as np\nimport matplotlib.pyplot as plt\n\nx = np.arange(1, 13)\ny = np.array([12, 14, 13, 18, 22, 30, 29, 24, 19, 15, 13, 12])\n\nfig, ax = plt.subplots(figsize=(7, 3.6))\nax.plot(x, y, marker=\"o\")\nax.axhline(y.mean(), ls=\"--\", color=\"gray\")\ni = y.argmax()\nax.annotate(\"peak\", xy=(x[i], y[i]), xytext=(x[i] + 1, y[i] + 2), arrowprops=dict(arrowstyle=\"->\"))\nax.set_title(\"Monthly demand\")\nax.set_xlabel(\"month\")\nax.set_ylabel(\"units\")"
  },
  {
   "id": "mp-3-2",
   "title": "A 2x2 dashboard",
   "tests": "fig = plt.gcf()\ncheck(len(fig.axes) == 4, 'the figure has 4 axes', 'plt.subplots(2, 2)')\na = fig.axes\ncheck(len(a[0].lines) >= 1 and len(a[1].patches) == 3 and len(a[2].patches) >= 5 and len(a[3].collections) >= 1, 'line, bar, histogram and scatter are in the right places', 'axes[0,0] line, [0,1] bar, [1,0] hist, [1,1] scatter')\ncheck(all(x.get_title() for x in a), 'every chart has a title', 'set_title on each axes')\ncheck(fig._suptitle is not None and fig._suptitle.get_text() == 'Dashboard', 'the figure title is \"Dashboard\"', 'fig.suptitle(\"Dashboard\")')",
   "hints": [
    "axes is a 2x2 array: axes[0, 0], axes[0, 1], axes[1, 0], axes[1, 1].",
    "fig.suptitle sets the overall title."
   ],
   "solution": "import numpy as np\nimport matplotlib.pyplot as plt\n\nx = np.arange(10)\ny = x ** 1.5\ncats, vals = [\"a\", \"b\", \"c\"], [5, 8, 3]\ndata = np.random.default_rng(2).normal(size=200)\n\nfig, axes = plt.subplots(2, 2, figsize=(8, 6))\naxes[0, 0].plot(x, y); axes[0, 0].set_title(\"line\")\naxes[0, 1].bar(cats, vals); axes[0, 1].set_title(\"bar\")\naxes[1, 0].hist(data, bins=15); axes[1, 0].set_title(\"histogram\")\naxes[1, 1].scatter(x, y); axes[1, 1].set_title(\"scatter\")\nfig.suptitle(\"Dashboard\")\nfig.tight_layout()"
  },
  {
   "id": "mp-4-1",
   "title": "A 3-D scatter",
   "tests": "ax = plt.gcf().axes[0]\ncheck(getattr(ax, 'name', '') == '3d', 'the axes is 3-D', 'add_subplot(projection=\"3d\")')\ncheck(len(ax.collections) == 1 and len(ax.collections[0]._offsets3d[0]) == 60, 'the 60 points are plotted in 3-D', 'ax.scatter(x, y, z, ...)')\ncheck(ax.get_xlabel() and ax.get_ylabel() and ax.get_zlabel(), 'all three axes are labelled', 'set_xlabel / set_ylabel / set_zlabel')",
   "hints": [
    "projection=\"3d\" makes a 3-D axes.",
    "scatter takes three coordinate arrays."
   ],
   "solution": "import numpy as np\nimport matplotlib.pyplot as plt\n\nrng = np.random.default_rng(4)\nx, y = rng.normal(size=60), rng.normal(size=60)\nz = x * 0.8 + y * 0.5 + rng.normal(0, 0.2, 60)\n\nfig = plt.figure(figsize=(6, 4.6))\nax = fig.add_subplot(projection=\"3d\")\nax.scatter(x, y, z, c=z, cmap=\"viridis\")\nax.set_xlabel(\"x\")\nax.set_ylabel(\"y\")\nax.set_zlabel(\"z\")"
  },
  {
   "id": "mp-5-1",
   "title": "Fix the bad chart",
   "tests": "ax = plt.gcf().axes[0]\nw = [p.get_width() for p in ax.patches]\ncheck(len(ax.patches) == 6 and sorted(w) == [60, 75, 95, 120, 140, 210], 'six horizontal bars with the right values', 'ax.barh(names, vals)')\ncheck(w == sorted(w), 'the largest bar is at the top (values ascend in drawing order)', 'sort smallest to largest before barh')\ncheck(len(ax.get_title()) > 3 and ax.get_xlabel() == 'units sold', 'there is a title and the x label is \"units sold\"', 'set_title and set_xlabel')\ncheck(len(ax.texts) >= 6, 'each bar has its value written on it', 'ax.bar_label(bars)')",
   "hints": [
    "In a barh chart the FIRST bar is drawn at the bottom, so feed the values sorted from smallest to largest.",
    "sorted(zip(values, names)) sorts the pairs by value.",
    "bars = ax.barh(...) returns the bars; ax.bar_label(bars) writes the numbers."
   ],
   "solution": "import matplotlib.pyplot as plt\n\nproducts = [\"Kettle\", \"Lamp\", \"Desk\", \"Chair\", \"Shelf\", \"Mat\"]\nsold = [140, 95, 60, 210, 75, 120]\n\npairs = sorted(zip(sold, products))          # smallest first, so the largest ends up at the top of a barh\nvals = [p[0] for p in pairs]\nnames = [p[1] for p in pairs]\n\nfig, ax = plt.subplots(figsize=(6, 3.8))\nbars = ax.barh(names, vals, color=\"#2F6FE0\")\nax.bar_label(bars, padding=3)\nax.set_title(\"Chair is our best seller\")\nax.set_xlabel(\"units sold\")"
  },
  {
   "id": "mp-5-2",
   "title": "Seaborn in one line",
   "tests": "ax = plt.gcf().axes[0]\ncheck(len(ax.patches) >= 3, 'three boxes are drawn', 'sns.boxplot(...)')\ncheck(ax.get_title() == 'Waiting time by department', 'the title is right', 'ax.set_title(\"Waiting time by department\")')\ncheck(ax.get_ylabel() == 'wait' and ax.get_xlabel() == 'dept', 'seaborn labelled the axes from the column names', 'x=\"dept\", y=\"wait\"')",
   "hints": [
    "sns.boxplot(data=df, x=..., y=..., ax=ax)"
   ],
   "solution": "import numpy as np\nimport pandas as pd\nimport seaborn as sns\nimport matplotlib.pyplot as plt\n\nrng = np.random.default_rng(1)\ndf = pd.DataFrame({\"dept\": np.repeat([\"ENT\", \"General\", \"Cardio\"], 60),\n                   \"wait\": np.concatenate([rng.normal(15, 4, 60), rng.normal(30, 9, 60), rng.normal(22, 6, 60)])})\nfig, ax = plt.subplots(figsize=(6, 3.8))\nsns.boxplot(data=df, x=\"dept\", y=\"wait\", ax=ax)\nax.set_title(\"Waiting time by department\")"
  },
  {
   "id": "mp-6-1",
   "title": "Draw a correlation heatmap",
   "tests": "fig = plt.gcf(); ax = fig.axes[0]\ncheck(im.get_cmap().name == 'coolwarm', 'diverging colormap coolwarm', 'cmap=\"coolwarm\"')\ncheck(tuple(im.get_clim()) == (-1, 1), 'the colour scale is fixed from -1 to 1', 'vmin=-1, vmax=1')\ncheck(np.allclose(np.asarray(im.get_array()), corr.values), 'the image shows the correlation matrix', 'ax.imshow(corr.values, ...)')\ncheck([t.get_text() for t in ax.get_xticklabels()] == ['a', 'b', 'c'] and len(fig.axes) == 2, 'x labels a, b, c and a colorbar axis exist', 'set_xticklabels, fig.colorbar(im)')",
   "hints": [
    "A correlation lives between -1 and 1; a diverging colormap is coolwarm."
   ],
   "solution": "import numpy as np\nimport pandas as pd\nimport matplotlib.pyplot as plt\n\nrng = np.random.default_rng(0)\na = rng.normal(size=100)\ndf = pd.DataFrame({\"a\": a, \"b\": a * 0.8 + rng.normal(0, 0.5, 100), \"c\": rng.normal(size=100)})\ncorr = df.corr()\nfig, ax = plt.subplots()\nim = ax.imshow(corr.values, cmap=\"coolwarm\", vmin=-1, vmax=1)\nax.set_xticks(range(3))\nax.set_xticklabels(corr.columns)\nfig.colorbar(im)"
  },
  {
   "id": "mp-6-2",
   "title": "Frames of an animation",
   "tests": "fig = plt.gcf()\ncheck(len(fig.axes) == 4, 'four axes', 'plt.subplots(1, 4)')\ncheck([len(a_.lines[0].get_xdata()) for a_ in fig.axes] == [30, 60, 90, 120], 'the axes show 30, 60, 90 and 120 points', 'x[:frame]')\ncheck([a_.get_title() for a_ in fig.axes] == ['frame 30', 'frame 60', 'frame 90', 'frame 120'], 'titles frame 30 ... frame 120', 'f\"frame {frame}\"')",
   "hints": [
    "Inside the loop the variable is called frame."
   ],
   "solution": "import numpy as np\nimport matplotlib.pyplot as plt\n\nx = np.linspace(0, 6.28, 120)\ny = np.sin(x)\nfig, axes = plt.subplots(1, 4, figsize=(10, 2.5), sharey=True)\nfor ax, frame in zip(axes, [30, 60, 90, 120]):\n    ax.plot(x[:frame], y[:frame])\n    ax.set_title(f\"frame {frame}\")\n    ax.set_xlim(0, 6.28)"
  }
 ]
})
```
